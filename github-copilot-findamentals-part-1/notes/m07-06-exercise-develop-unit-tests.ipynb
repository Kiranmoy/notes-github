# Exercise - Develop unit tests using GitHub Copilot

**Important**
* To complete this exercise, you need an active GitHub account and a Visual Studio Code environment configured for C# development.
* For help with configuring your lab environment, see **[Configure lab environment for GitHub Copilot Labs](https://go.microsoft.com/fwlink/?linkid=2320147)**.
* For help with enabling GitHub Copilot in Visual Studio Code, see **[Enable GitHub Copilot in Visual Studio Code](https://go.microsoft.com/fwlink/?linkid=2320147)**.

In this exercise, you use GitHub Copilot to accelerate the development of unit tests for a C# application. Tasks include:
* Use the Chat view and Agent to create a new test class.
* Use the Chat view and Agent mode to create unit tests.

When you select the launch exercise button, the browser navigates to a public GitHub page that provides instructions for this exercise.

When you finish the exercise, return here for:
* A quick knowledge check.
* A summary of what you learned during this module.
* A badge for completing this module.
